## Time delay
$$y(t) = x(t-t_d)$$

$t_d$ is delay time

Need to assume $y(t)$ is $0$ before shift starts (cannot know the future)

$$y(t) = x(t-t_d)u(t-t_d)$$
$u(t)$ is the unit step function

$$Y(s) = \mathcal{L}\left\{y(t)\right\} = \mathcal{L}\left\{x(t-t_d)u(t-t_d)\right\}$$

$$f(t-a)u(t-a) \xLeftrightarrow{\mathcal{L}} e^{-as}F(s) $$

$$Y(s) = e^{-t_ds}X(s)$$

$$H_\text{delay}(s) = \frac{e^{-t_ds}X(s)}{X(s)} = e^{-t_ds}$$
